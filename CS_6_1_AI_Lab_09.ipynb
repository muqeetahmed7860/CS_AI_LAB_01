Import Libraries


```python
import numpy as np
import pandas as pd
from sklearn.model_selection import train_test_split
from sklearn.neighbors import KNeighborsClassifier
from sklearn.metrics import accuracy_score
from sklearn.datasets import load_iris
```

Load Dataset

```python
data = load_iris()
X = data.data
y = data.target

print("Features Shape:", X.shape)
print("Labels Shape:", y.shape)
```

Split Dataset

```python
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.2)

print("Training Data:", X_train.shape)
print("Testing Data:", X_test.shape)
```

Apply KNN with Different K


```python
k_values = [1, 3, 5, 7, 9]

for k in k_values:
    model = KNeighborsClassifier(n_neighbors=k)
    model.fit(X_train, y_train)

    y_pred = model.predict(X_test)

    acc = accuracy_score(y_test, y_pred)
    print(f"K = {k}, Accuracy = {acc}")
```

Second Dataset (Diabetes)

```python
from sklearn.datasets import load_diabetes

data2 = load_diabetes()
X2 = data2.data
y2 = data2.target

X_train2, X_test2, y_train2, y_test2 = train_test_split(X2, y2, test_size=0.2)

for k in [3,5,7]:
    model2 = KNeighborsClassifier(n_neighbors=k)
    model2.fit(X_train2, y_train2)

    y_pred2 = model2.predict(X_test2)

    acc2 = accuracy_score(y_test2, y_pred2.round())
    print(f"[Diabetes] K = {k}, Accuracy = {acc2}")
```

Final Result Conclusion

```python
print("Conclusion:")
print("Different K values give different accuracy.")
print("We select the K with highest accuracy.")
```
